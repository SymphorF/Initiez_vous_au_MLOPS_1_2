### I- Création du fichier de production fictif pour tester le drift

```python
import pandas as pd

# Dataset complet fourni
data = pd.read_csv("fichier_test.csv")

# Simulation :
train_data = data.sample(frac=0.7, random_state=42)  # données d'entraînement
prod_data = data.sample(frac=0.3, random_state=24)   # "fausse" production

# Sauvegarde des fichiers CSV
train_data.to_csv("train_data.csv", index=False)
prod_data.to_csv("prod_data.csv", index=False)
```
