```python
import pandas as pd
from evidently import Report
from evidently.presets import DataDriftPreset

# Chargement des datasets
reference = pd.read_csv("train_data.csv")  # données d'entraînement
production = pd.read_csv("prod_data.csv")  # données de "production"

# Création du rapport de drift
report = Report(metrics=[DataDriftPreset(method="psi")],include_tests="True")
report_drift = report.run(current_data=production, reference_data=reference)

# Sauvegarde du rapport au format HTML
report_drift.save_html("report_drift.html")
```

***Ouvrir le fichier html téléchargé pour visualiser le rapport***

**Drift score / test statistique :**

Si psi (Population Stability Index) est utilisé :

< 0.1 → stable

0.1–0.25 → léger drift

0.25 → drift important

| **Niveau de drift**      | **Description**                                                       | **Exemple numérique**            | **Implications pour le modèle**                                                                   |
| ------------------------ | --------------------------------------------------------------------- | -------------------------------- | ------------------------------------------------------------------------------------------------- |
| **Stable / Aucun drift** | La distribution en production est très proche de celle d’entraînement | Moyenne `revenu` : 3000€ → 3050€ | Le modèle fonctionne normalement, pas d’action requise                                            |
| **Drift léger**          | Différence détectable mais faible                                     | Moyenne `âge` : 35 → 38 ans      | Le modèle pourrait légèrement moins bien prédire, surveiller les tendances                        |
| **Drift important**      | La distribution en production est très différente                     | Moyenne `revenu` : 3000€ → 6000€ | Risque de prédictions incorrectes ou biaisées, envisager recalibrage ou ré-entraînement du modèle |
